```python
data = data[data["weight"] >0 ]
```

```python
data["rate_per_mile"] = data["posted_rate"] / data["distance"]
```

```python
Q1 = data["rate_per_mile"].quantile(0.25)
Q3 = data["rate_per_mile"].quantile(0.75)
IQR = Q3 - Q1
lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR

before = len(data)
data = data[(data["rate_per_mile"] >= lower) & (data["rate_per_mile"] <= upper)]
print(f"Dropped {before - len(data)} outlier rows")
```

```python
data = data.drop("rate_per_mile", axis=1)
```

train test split here V.V.V

```python
from sklearn.model_selection import train_test_split
```

```python
x = data.drop("posted_rate",axis=1)
y = data["posted_rate"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
x_train.info()
```

```python
cat_cols = x.select_dtypes(include='object').columns.tolist()
num_cols = x.select_dtypes(include='number').columns.tolist()
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer([
    ("cat",OneHotEncoder(handle_unknown="ignore",sparse_output=False),cat_cols),
    ("num",StandardScaler(),num_cols)
])
```

```python

X_transformed = preprocessor.fit_transform(x_train)

df = pd.DataFrame(
    X_transformed,
    columns=preprocessor.get_feature_names_out()
)
```

```python
df.head()
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

```python
rf_model = RandomForestRegressor(n_estimators=100,n_jobs=-1,random_state=42)
```

```python
rf_model.fit(X_transformed,y_train)
```

```python
transformed_test =  preprocessor.transform(x_test)
```

```python
y_pred = rf_model.predict(transformed_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
print(mean_squared_error(y_test,y_pred))
```

```python

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
# print("RMSE:", root_mean_squared_error(y_test,y_pred))
```

```python
# 1. Feature importance
importances = pd.Series(
    rf_model.feature_importances_,
    index=preprocessor.get_feature_names_out()
).sort_values(ascending=False)
print(importances.head(10))

# 2. Cross-validation
from sklearn.model_selection import cross_val_score
cv_scores = cross_val_score(rf_model, X_transformed, y_train, cv=5, scoring="r2")
print("CV R2 scores:", cv_scores)
print("Mean:", cv_scores.mean(), "Std:", cv_scores.std())
```

```python
preprocessor_final = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols),
    ("num", StandardScaler(), num_cols)
])

X_full_transformed = preprocessor_final.fit_transform(x)

final_model = RandomForestRegressor(n_estimators=100, n_jobs=-1, random_state=42)
final_model.fit(X_full_transformed, y)

```
